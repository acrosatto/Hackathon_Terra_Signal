```python
%sql
-- ============================================================
-- Tabela terra_signal_history_intelligence (Terra Signal, history)
-- ============================================================
-- O que faz: carrega o arquivo CSV da Terra Signal (customer intelligence, history) em tabela.
-- Origem:    terra_signal_history_intelligence.csv (volume bases, schema bases)
-- Saída:     tabela terra_signal_history_intelligence, com as colunas do CSV, todas como texto.
-- Regras:    separador do CSV ";" e cabeçalho na primeira linha. Os tipos não são inferidos. É a referência do colega para a terra_signal_history_intelligence_Adri.
-- ============================================================

CREATE OR REPLACE TABLE hackaton_databricks.tabelas_hackaton.terra_signal_history_intelligence AS
SELECT *
FROM read_files(
  '/Volumes/hackaton_databricks/bases/bases/terra_signal_history_intelligence.csv',
  format => 'csv',
  sep => ';',
  header => true,
  inferColumnTypes => false
);

SELECT COUNT(*) AS linhas
FROM hackaton_databricks.tabelas_hackaton.terra_signal_history_intelligence;

SELECT *
FROM hackaton_databricks.tabelas_hackaton.terra_signal_history_intelligence
LIMIT ALL;
```

```python
%sql
-- ============================================================
-- Validação 13: limites do SentimentoVADER no colega (Terra Signal, history)
-- ============================================================
-- O que faz: mostra o menor e o maior SentimentScore de cada SentimentoVADER na tabela do colega, para conferir o limite de ±0,05.
-- Origem:    terra_signal_history_intelligence
-- Saída:     consulta (SentimentoVADER, score_min, score_max, n), sem criar tabela.
-- Regras:    o score vem como texto com vírgula decimal, então a vírgula é trocada por ponto antes da conversão. Linhas sem rótulo aparecem como nulo.
-- ============================================================
  SELECT SentimentoVADER,
         MIN(try_cast(replace(SentimentScore, ',', '.') AS DOUBLE)) AS score_min,
         MAX(try_cast(replace(SentimentScore, ',', '.') AS DOUBLE)) AS score_max,
         COUNT(*) AS n
  FROM hackaton_databricks.tabelas_hackaton.terra_signal_history_intelligence
  GROUP BY SentimentoVADER
```
